# Calculating density of fiberglass used in subscale rocket from mass and volume

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import scipy.interpolate as interpolate

plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['figure.titlesize'] = 20
plt.rcParams['figure.titleweight'] = 'bold'
plt.rcParams['figure.figsize'] = (10, 8)
plt.rcParams['axes.titlesize'] = 16
plt.rcParams['axes.labelsize'] = 14
plt.rcParams['figure.labelsize'] = 14
plt.rcParams['xtick.labelsize'] = 12
plt.rcParams['ytick.labelsize'] = 12
plt.rcParams['legend.fontsize'] = 12
plt.rcParams['mathtext.fontset'] = 'cm'
```

### Using forward body tube

```python
# Dimensions of forward body tube
forward_mass = 324.3 / 453.592 # lb
forward_thickness = 1.1 / 25.4 # in
forward_length = 21 + 15/16 # in
forward_ID = 4 # in

print('Forward Body Tube:')
print(f'  Mass: {forward_mass:.3f} lb')
print(f'  Thickness: {forward_thickness:.3f} in')
print(f'  Length: {forward_length:.3f} in')

# Number of holes in forward body tube
num_retention_holes = 6
retention_hole_diameter = 17/64 # in
num_nose_cone_holes = 2
nose_cone_hole_diameter = 1/8 # in
```

```python
# Volume of holes
retention_hole_volume = num_retention_holes * np.pi * (retention_hole_diameter / 2)**2 * forward_thickness # in^3
nose_cone_hole_volume = num_nose_cone_holes * np.pi * (nose_cone_hole_diameter / 2)**2 * forward_thickness # in^3
total_hole_volume = retention_hole_volume + nose_cone_hole_volume
print("Total hole volume (in^3): ", total_hole_volume)

# Volume of forward body tube
forward_outer_diameter = forward_ID + forward_thickness # in
forward_volume = (np.pi / 4) * (forward_outer_diameter**2 - forward_ID**2) * forward_length # in^3
print("Forward body tube volume (in^3): ", forward_volume)

# Actual material volume
actual_material_volume = forward_volume - total_hole_volume
print("Actual material volume (in^3): ", actual_material_volume)
# Material density
forward_material_density = forward_mass / actual_material_volume # lb/in^3
print("Material density (lb/in^3): ", forward_material_density)
```

### Using middle body tube

```python
# Dimensions of forward body tube
middle_mass = 320.0 / 453.592 # lb
middle_thickness = 1.1 / 25.4 # in
middle_length = 21 + 1/16 # in
middle_ID = 4 # in

print('Middle Body Tube:')
print(f'  Mass: {middle_mass:.3f} lb')
print(f'  Thickness: {middle_thickness:.3f} in')
print(f'  Length: {middle_length:.3f} in')

# Number of holes in middle body tube
num_retention_holes = 4
retention_hole_diameter = 17/64 # in
num_shear_pin_holes = 3
shear_pin_hole_diameter = 5/32 # in
```

```python
# Volume of holes
retention_hole_volume = num_retention_holes * np.pi * (retention_hole_diameter / 2)**2 * middle_thickness # in^3
shear_pin_hole_volume = num_shear_pin_holes * np.pi * (shear_pin_hole_diameter / 2)**2 * middle_thickness # in^3
total_hole_volume = retention_hole_volume + shear_pin_hole_volume
print("Total hole volume (in^3): ", total_hole_volume)

# Volume of middle body tube
middle_outer_diameter = middle_ID + middle_thickness # in
middle_volume = (np.pi / 4) * (middle_outer_diameter**2 - middle_ID**2) * middle_length # in^3
print("Middle body tube volume (in^3): ", middle_volume)

# Actual material volume
actual_material_volume = middle_volume - total_hole_volume
print("Actual material volume (in^3): ", actual_material_volume)
# Material density
middle_material_density = middle_mass / actual_material_volume # lb/in^3
print("Material density (lb/in^3): ", middle_material_density)
```

```python
# Average material density
average_material_density = (forward_material_density + middle_material_density) / 2 # lb/in^3
print("Average material density (lb/in^3): ", average_material_density)
```

### Defining full scale forward body tube parameters

```python
forward_ID = 6 # in
forward_thickness = 0.12 # in
forward_length = 33 # in

# Volume of body tube
forward_outer_diameter = forward_ID + forward_thickness # in
forward_volume = (np.pi / 4) * (forward_outer_diameter**2 - forward_ID**2) * forward_length # in^3
print("Full scale forward body tube volume (in^3): ", forward_volume)

# Mass of body tube
forward_mass = average_material_density * forward_volume # lbs
print("Full scale forward body tube mass (lb): ", forward_mass)
```
